```python
r=int(input("Enter the number of rats:"))
unit=int(input("Enter food needed by each rat:"))
n=int(input("Enter number of houses:"))
ar=list(map(int,input("Enter food in each house:").split()))
food=r*unit
total=0
houses=0
if len(ar)==0:
  print(-1)
else:
  for i in range(n):
    total=total+ar[i]
    houses=houses+1
    if total>=food:
      break
  if total<food:
    print(0)
  else:
    print(houses)
```

Reverse a number

```python
n=int(input("Enter the number:"))
opt=0
while(n>0):
  rem=n%10
  opt=opt*10+rem
  n=n//10
print("Reversed number=",opt)
```

Sum of the digits of a number

```python
n=int(input("Enter the number:"))
sum=0
while(n>0):
  rem=n%10
  sum=sum+rem
  n=n//10
print("Sum=",sum)
```

Largest digit in a number

```python
n=int(input("Enter the number:"))
lrgt=0
while n>0:
  num=n%10
  if num>lrgt:
    lrgt=num
  n=n//10
print("Largest digit=",lrgt)
```

String palindrome

```python
n=input("Enter the string:")
if n==n[::-1]:
  print(n,"is a palindrome")
else:
  print(n,"is not a palindrome")
```

Number palindrome

```python
n=int(input("Enter the number:"))
temp=n
opt=0
while n>0:
  num=n%10
  opt=opt*10+num
  n=n//10
if(temp==opt):
  print(temp,"is a palindrome")
else:
  print(temp,"is not a palindrome")
```

Bridge

```python
n1=int(input("Enter the number:"))
n2=int(input("Enter the number:"))
a=[]
b=[]
for i in range(n1):
  a.append(int(input("Enter the element:")))
for i in range(n2):
  b.append(int(input("Enter the element:")))
c=list(set(a+b))
c.sort()
n=len(c)
if n%2==1:
  print(n//2)
else:
  print((c[n//2-1]+c[n//2])/2)
```
